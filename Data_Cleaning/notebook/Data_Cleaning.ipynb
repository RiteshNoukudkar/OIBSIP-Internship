```python
plt.figure(figsize=(10,6))

sns.heatmap(df.isnull(), cbar=False, cmap="viridis")

plt.title("Missing Values Heatmap")
plt.savefig("../images/missing_values_heatmap.png", dpi=300, bbox_inches="tight")
plt.show()
```

### Observation

- No missing values remain after preprocessing.
- The dataset is now complete and ready for analysis.

## Income Distribution

```python
plt.figure(figsize=(8,5))

sns.histplot(df["Income"], bins=20, kde=True)

plt.title("Income Distribution")
plt.xlabel("Income")
plt.ylabel("Count")

plt.show()
```

### Observation

- Most customers fall within the middle-income range.
- A few customers have significantly higher incomes.

### Outlier Detection (Boxplot)

```python
plt.figure(figsize=(8,5))

sns.boxplot(x=df["Income"])

plt.title("Income Boxplot")

plt.show()
```

### Observation

- The Income column contains a few outliers.
- These represent customers with exceptionally high income.

### Correlation Heatmap

```python
plt.figure(figsize=(14,10))

sns.heatmap(
    df.select_dtypes(include="number").corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()
```
